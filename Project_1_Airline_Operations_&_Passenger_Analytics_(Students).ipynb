## Q15. Passenger Age Groups

Create suitable age groups.

For example:

- Young
- Adult
- Senior

Calculate the number of passengers in each group.

**Your code:**

```python
# Write your code here
```

```python
bins = [0, 30, 60, 100]

labels = [
    "Young",
    "Adult",
    "Senior"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

age_group_count = (
    df["Age Group"]
    .value_counts()
    .sort_index()
)

print(age_group_count)
```

# SECTION 6 — SATISFACTION ANALYSIS

## Q21. Overall Passenger Satisfaction

Find:

- Satisfied passengers
- Neutral/dissatisfied passengers
- Overall satisfaction percentage

**Your code:**

```python
satisfaction_count = (
    df["satisfaction"]
    .value_counts()
)

print("Satisfaction Count:")
print(satisfaction_count)
```

```python
satisfied_count = (
    df["satisfaction"] == "satisfied"
).sum()

neutral_dissatisfied_count = (
    df["satisfaction"] != "satisfied"
).sum()

satisfaction_percentage = (
    satisfied_count / len(df)
) * 100

print("Satisfied passengers:",
      satisfied_count)

print("Neutral/Dissatisfied passengers:",
      neutral_dissatisfied_count)

print("Overall satisfaction percentage:",
      round(satisfaction_percentage, 2), "%")
```

## Q22. Satisfaction by Class

Calculate the satisfaction percentage for:

- Business
- Eco
- Eco Plus

**Your code:**

```python
# Write your code here
```

```python
satisfaction_by_class = (
    df.groupby("Class")["satisfaction"]
    .apply(
        lambda x: (x == "satisfied").mean() * 100
    )
)

print(
    satisfaction_by_class.round(2)
)
```

## Q23. Satisfaction by Customer Type

Compare satisfaction between:

- Loyal Customers
- Disloyal Customers

Calculate the satisfaction percentage for each.

**Your code:**

```python
satisfaction_by_customer = (
    df.groupby("Customer Type")["satisfaction"]
    .apply(
        lambda x: (x == "satisfied").mean() * 100
    )
)

print(
    satisfaction_by_customer.round(2)
)
```

## Q25. Delay vs Passenger Satisfaction

Create a new column:

`Departure_Delay_Status`

with:

- Delayed → Departure Delay > 0
- Not Delayed → Departure Delay = 0

Compare passenger satisfaction between the two groups.

### Business Question

**Does passenger satisfaction differ between delayed and non-delayed flights in this dataset?**

**Your code:**

```python
df["Departure_Delay_Status"] = np.where(
    df["Departure Delay in Minutes"] > 0,
    "Delayed",
    "Not Delayed"
)

print(
    df["Departure_Delay_Status"]
    .value_counts()
)
```

```python
delay_satisfaction = (
    df.groupby("Departure_Delay_Status")["satisfaction"]
    .apply(
        lambda x: (x == "satisfied").mean() * 100
    )
)

print(
    delay_satisfaction.round(2)
)
```
